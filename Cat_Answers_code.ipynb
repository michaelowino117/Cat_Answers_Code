```python
#Question 1(b) python program
import heapq

def a_star(graph, start, goal, heuristic):
    open_list = []
    heapq.heappush(open_list, (0, start))

    g_cost = {start: 0}
    parent = {start: None}

    while open_list:
        _, current = heapq.heappop(open_list)

        if current == goal:
            path = []
            while current:
                path.append(current)
                current = parent[current]
            return path[::-1]

        for neighbor, cost in graph[current]:
            new_g = g_cost[current] + cost

            if neighbor not in g_cost or new_g < g_cost[neighbor]:
                g_cost[neighbor] = new_g
                f_cost = new_g + heuristic[neighbor]
                heapq.heappush(open_list, (f_cost, neighbor))
                parent[neighbor] = current

    return None


# Example usage
graph = {
    'A': [('B', 1), ('C', 4)],
    'B': [('D', 2), ('E', 5)],
    'C': [('F', 1)],
    'D': [],
    'E': [('F', 1)],
    'F': []
}

heuristic = {
    'A': 7,
    'B': 6,
    'C': 2,
    'D': 1,
    'E': 1,
    'F': 0
}

path = a_star(graph, 'A', 'F', heuristic)
print("Optimal Path:", path)
```

```python
# Vacuum Cleaner Agent Simulation
# Two locations: A and B
# Each location can be Dirty or Clean

class VacuumCleanerAgent:
    def __init__(self):
        self.location = 'A'   # starting position
        self.environment = {
            'A': 'Dirty',
            'B': 'Dirty'
        }

    def perceive(self):
        # check current location status
        return self.environment[self.location]

    def clean(self):
        print(f"Cleaning location {self.location}")
        self.environment[self.location] = 'Clean'

    def move(self):
        # move between A and B
        if self.location == 'A':
            self.location = 'B'
        else:
            self.location = 'A'
        print(f"Moved to {self.location}")

    def run(self):
        # stop when both rooms are clean
        while self.environment['A'] == 'Dirty' or self.environment['B'] == 'Dirty':

            status = self.perceive()

            if status == 'Dirty':
                self.clean()
            else:
                self.move()

        print("All locations are clean!")


# Run the agent
agent = VacuumCleanerAgent()
agent.run()
```
